### Preprocessing Pipeline Overview
Scalp electroencephalography recordings contain physiological contaminants (ocular blinks, cardiac pulses) and environmental noise (mains power hum, high-frequency muscle contractions). 

This notebook implements an automated, scientifically validated signal conditioning pipeline:
1. **Digital Frequency Filtering:** Cascaded zero-phase Butterworth 4th-order bandpass ($8 - 32\text{ Hz}$) and $50\text{ Hz}$ Notch filter ($Q = 30$).
2. **Artifact Attenuation (FastICA):** Decompose signals into independent components, identify components correlated with $4\times\text{EOG}$ sensors, and project out ocular artifacts.
3. **Bad Channel Interpolation:** Spherical spline interpolation of disconnected or anomalous electrodes.
4. **Spatial Re-Referencing:** Common Average Reference (CAR) to suppress global volume-conducted noise.
5. **Temporal Segmentation & Decimation:** Crop active motor imagery interval ($[1.5, 4.5]\text{ s}$) and downsample from $1000\text{ Hz}$ to $250\text{ Hz}$ ($4\times$ memory reduction).
6. **Cross-Validation Partition Generation:** Generate deterministic, leakage-free indices for Within-Session, Cross-Session, and Cross-Subject (LOSO) benchmarks.
