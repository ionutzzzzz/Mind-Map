### Neurophysiological Foundation: Sensorimotor Rhythms (SMR)
Motor imagery (mental rehearsal of limb movements without overt physical execution) elicits stereotypic, localized oscillatory modulations within the primary motor cortex ($M1$) and primary somatosensory cortex ($S1$):

* **Mu Rhythm ($8 - 12\text{ Hz}$):** Generated by thalamocortical networks during physical stillness. Imagining movement of the contralateral hand induces **Event-Related Desynchronization (ERD)**—a sharp drop in spectral power over the contralateral motor cortex:
  * **Right Hand Imagery:** Induces Mu ERD over **C3** (left hemisphere).
  * **Left Hand Imagery:** Induces Mu ERD over **C4** (right hemisphere).
* **Beta Rhythm ($16 - 24\text{ Hz}$):** Modulates during active movement and motor inhibition. Desynchronizes during the motor task and exhibits a synchronized power rebound (**ERS**) following movement termination.

### The Pfurtscheller ERD/ERS Formulation:
$$\text{ERD/ERS}\%(f, t) = \frac{A(f, t) - R(f)}{R(f)} \times 100\%$$

Where $A(f, t)$ is the spectral power during active imagery ($1.5 - 5.5\text{ s}$), and $R(f)$ is the resting baseline power ($-1.0 - 0.0\text{ s}$). Negative values denote desynchronization (cortical activation); positive values denote synchronization (cortical idling).
