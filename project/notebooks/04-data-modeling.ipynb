### Modeling Framework & Evaluation Objectives
This notebook delivers the core comparative benchmark and hardware adaptation engine:
1. **Classical Spatial & Riemannian Pipelines:**
   * Common Spatial Pattern (CSP) + Shrinkage Linear Discriminant Analysis (LDA).
   * Filter Bank CSP (FBCSP) + Mutual Information Best Individual Features (MIBIF) + SVM.
   * Covariance Matrix Tangent Space on Riemannian Manifolds + Ridge Classifier.
2. **Domain Adaptation:** Euclidean Alignment (EA) to enable zero-calibration cross-subject transfer learning.
3. **Deep Convolutional Neural Networks (PyTorch):**
   * **EEGNet:** Compact depthwise & separable convolutional network ($< 3,000$ parameters).
   * **DeepConvNet:** 4-stage representation learner.
   * **ShallowFBCSPNet:** Direct neural parameterization of bandpower energy.
4. **Validation Protocols:** Within-Session (10-fold CV), Cross-Session (inter-day), and Cross-Subject (Leave-One-Subject-Out / LOSO).
5. **Wearable Channel Reduction:** Systematic channel pruning ($59 \rightarrow 32 \rightarrow 16 \rightarrow 8 \rightarrow 3$ channels) mapping directly to the **OpenBCI Ultracortex Mark IV** headset.
6. **Embedded Model Serialization:** Export champion 16-channel model to ONNX runtime for real-time inference with BrainFlow.
